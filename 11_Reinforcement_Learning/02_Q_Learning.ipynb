```python
import gymnasium as gym
import numpy as np
import random
```

```python
gamma = 0.99
alpha = 0.5
epsilon = 0.1
episodes = 500
```

```python
# Q-table  => store Q-values
Q = np.zeros((48,4))
```

```python
# Epsilon- policy

def epsilon_greedy(state):
    if random.random() < epsilon :
        return env.action_space.sample() # explore
    else:
        return np.argmax(Q[state])
```

```python
# Q-learning

for episode in range(episodes):
    # render = (episode % 50 == 0) 
    # if render:
    #     env=gym.make("CliffWalking-v1",render_mode = "human")
    #     print(f"rendering env for episode={episode+1}/500")
    # else :
    #      env=gym.make("CliffWalking-v1")all
    env=gym.make("CliffWalking-v1")
    done = False
    state,_ = env.reset()

    episode_len = 0
    total_reward = 0

    while not done:
        action = epsilon_greedy(state)
        next_state , reward , terminated , truncated ,_ = env.step(action)
        done = terminated or truncated 


        # Q-learning update 
        Q[state,action] += alpha * (
            reward + gamma * np.max(Q[next_state]) - Q[state,action] 
        )

        state = next_state
        
        total_reward += reward
        episode_len += 1

    print(f" episode = {episode+1}/500: total reward = {total_reward} & episode length = {episode_len}")
    env.close()
```

```python
# What did our agent learn ?

env=gym.make("CliffWalking-v1",render_mode = "human")
state , _ = env.reset()
done = False # when the episode has ended 
total_reward = 0
episode_len = 0

while not done:
    action = np.argmax(Q[state])
    state,reward,terminated,truncated,_ = env.step(action)

    done = terminated or truncated
    
    total_reward += reward
    episode_len += 1

print(f" total reward = {total_reward} & episode length = {episode_len}")
env.close()
```
